Q1. Write a function that return the range (maximum and minimum) of a given list
of numbers


```python
def find_range(numbers):
  max_num= max(numbers)
  min_num = min(numbers)
  return max_num,min_num
```

```python
data = [9,7,66,6,4,3,22]
find_range(data)
```

Q3. Write a function that returns the median value of a list of numbers


```python
def find_median(nums):
  s = sorted(nums)
  n = len(s)
  mid = n//2
  if n%2 == 0:
    return (s[mid-1] + s[mid]) /2
  return s[mid]
```

```python
l2 = [44,9,78,56,44,335]
find_median(l2)
```

####Q9. Write function to compute and return Z-score values of a list of numbers


```python
def z_scores(numbers):
    mean = sum(numbers) / len(numbers)

    variance = sum((x - mean) ** 2 for x in numbers) / len(numbers)
    std = variance ** 0.5

    return [(x - mean) / std for x in numbers]

print(z_scores([10, 20, 30, 40, 50]))
```

####Q10. Write a function to compute and return the Pearson correlation coefficient


```python
 def pearson_correlation(x, y):
    mean_x = sum(x) / len(x)
    mean_y = sum(y) / len(y)

    numerator = sum((a - mean_x) * (b - mean_y) for a, b in zip(x, y))

    denominator = (
        sum((a - mean_x) ** 2 for a in x) *
        sum((b - mean_y) ** 2 for b in y)
    ) ** 0.5

    return numerator / denominator


print(pearson_correlation([1, 2, 3, 4, 5], [2, 4, 6, 8, 10]))
```
