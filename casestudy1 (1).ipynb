3. Password Validator

Write a function that checks if a password is strong: at least 8 characters,
contains at least one digit, one uppercase letter, and one special character
(!@#$%).


```python
password=input('Enter password')
dig=0
upper=0
symb=0
symbol='!@#$%'
if len(password)>=8:
  for i in password:
    if i.isdigit():
      dig=dig+1
    if i.isupper():
      upper=upper+1
    if i in symbol:
      symb=symb+1
  if dig>0 and upper>0 and symb>0:
    print('Strong Password')
  else:
    print('Weak password!')
else:
    print('weak password!!')
```

2. Weather Data Analyzer

a. Analyze weather data (temperature, humidity, etc.) for a given region
b. Use variables and data types to store weather date

c. Implement conditional statements to determine weather conditions
(sunny, rainy, etc.)

d. Use functions to calculate average temperature and humidity, and
print results




```python
region = input("Enter the region: ")

temperature = [37, 56, 31, 29, 33]
humidity = [70, 65, 72, 55, 35]

weather = input("Enter weather condition (sunny/rainy/cloudy): ")

def average_temperature():
    total = sum(temperature)
    average = total / len(temperature)
    return average

def average_humidity():
    total = sum(humidity)
    average = total / len(humidity)
    return average

print("\nWeather Report")
print("Region:", region)
print("Weather:", weather)

if weather == "sunny":
    print("It is a sunny day.")
elif weather == "rainy":
    print("It is a rainy day.")
elif weather == "cloudy":
    print("It is a cloudy day.")
else:
    print("Weather condition is not recognized.")

print("Average Temperature:", average_temperature())
print("Average Humidity:", average_humidity())
```

4. Word Frequency Counter

Given a paragraph of text, write a function that returns the 3 most
frequently occurring words (ignore case and punctuation).



```python
word = input("Enter a word: ")
letter_frequency = {}
for i in word:
    if i in letter_frequency:
        letter_frequency[i] += 1
    else:
        letter_frequency[i] = 1
print(letter_frequency)
# Sort by frequency
sorted_frequency = sorted(letter_frequency.items(), key=lambda x: x[1], reverse=True)
# Get top 3
top_3 = sorted_frequency[:3]
print("Top 3 most frequent letters:", top_3)
```
